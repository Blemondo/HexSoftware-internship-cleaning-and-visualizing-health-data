# Project 02 – Hospital Patient Data Analysis

Analyze 600 synthetic hospital records covering patient age, disease, admission/discharge times, departments, length of stay, weight, and outcomes.


## 1. Import libraries


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style='whitegrid')
```

## 2. Load and inspect


```python
df=pd.read_csv('hospital_patient_data_600.csv')
print(df.shape)
display(df.head())
df.info()
```

## 3. Clean the data


```python
print(df.isnull().sum())
print('Duplicates:',df.duplicated().sum())
df['Admission_Time']=pd.to_datetime(df['Admission_Time']); df['Discharge_Time']=pd.to_datetime(df['Discharge_Time']); df['Weight_kg']=df['Weight_kg'].fillna(df['Weight_kg'].median())
```

## 4. Basic analysis


```python
display(df[['Age','Length_of_Stay_Days','Weight_kg']].describe())
print('Average age:',round(df.Age.mean(),2))
print('Average stay:',round(df.Length_of_Stay_Days.mean(),2),'days')
```

## 5. Visualization 1 — Patients by disease


```python
df['Disease_Type'].value_counts().plot(kind='bar',figsize=(10,5)); plt.title('Patients by Disease Type'); plt.ylabel('Patients'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## 6. Visualization 2 — Age distribution


```python
plt.figure(figsize=(9,5)); sns.histplot(df['Age'],bins=20,kde=True); plt.title('Distribution of Patient Ages'); plt.xlabel('Age'); plt.tight_layout(); plt.show()
```

## 7. Visualization 3 — Length of stay by disease


```python
plt.figure(figsize=(10,5)); sns.boxplot(data=df,x='Disease_Type',y='Length_of_Stay_Days'); plt.title('Length of Stay by Disease Type'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## 8. Visualization 4 — Admissions by department


```python
dept=df['Department'].value_counts().sort_values(); plt.figure(figsize=(9,5)); dept.plot(kind='barh'); plt.title('Admissions by Department'); plt.xlabel('Patients'); plt.tight_layout(); plt.show()
```

## 9. Visualization 5 — Age vs length of stay


```python
plt.figure(figsize=(9,5)); sns.scatterplot(data=df,x='Age',y='Length_of_Stay_Days',hue='Outcome',alpha=.7); plt.title('Age vs Length of Hospital Stay'); plt.tight_layout(); plt.show()
```

## 10. Outcome analysis


```python
out=df['Outcome'].value_counts(); print(out); print((out/len(df)*100).round(2))
```
